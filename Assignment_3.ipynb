```python

# Alpha-Beta Pruning
# CS 627 - Artificial Intelligence
# By Aniruddha Rath

def alpha_beta(node, level, alpha, beta):
    # Indentation for organized printing
    indent = " " * (level - 1)

    # If the node is a leaf, return its value
    if isinstance(node, int):
        print(f"{indent}Leaf: {node}")
        return node

    # Odd levels are MAX nodes
    if level % 2 == 1:
        print(f"{indent}MAX node (alpha={alpha}, beta={beta})")

        # Initialize starting value to negative infinity
        best_value = float("-inf")

        for child in node:
            # Recursively evaluate the child nodes
            successor_val = alpha_beta(
                child,
                level + 1,
                alpha,
                beta
            )

            # Update to best value found so far
            best_value = max(best_value, successor_val)

            # Update alpha at MAX nodes
            alpha = max(alpha, best_value)

            # Prune if alpha is greater than or equal to beta
            if alpha >= beta:
                print(f"{indent}Pruning remaining MAX branches")
                break

        return best_value

    # Even levels are MIN nodes
    else:
        print(f"{indent}MIN node (alpha={alpha}, beta={beta})")

        # Initialize starting value to positive infinity
        best_value = float("inf")

        for child in node:
            # Recursively evaluate the child nodes
            successor_val = alpha_beta(
                child,
                level + 1,
                alpha,
                beta
            )

            # Update to best value found so far
            best_value = min(best_value, successor_val)

            # Update beta at MIN nodes
            beta = min(beta, best_value)

            # Prune if beta is less than or equal to alpha
            if beta <= alpha:
                print(f"{indent}Pruning remaining MIN branches")
                break

        return best_value


# Example leaf values (replace with your lab's values if different)
treeValues = [3, 5, 6, 9, 1, 2, 0, -1]

# Construct a three-level binary game tree
tree = [
    [
        [treeValues[0], treeValues[1]],
        [treeValues[2], treeValues[3]]
    ],
    [
        [treeValues[4], treeValues[5]],
        [treeValues[6], treeValues[7]]
    ]
]

# Run alpha-beta pruning from the root MAX node
print("Alpha-Beta Pruning Results")
print("--------------------------")

result = alpha_beta(
    tree,
    1,
    float("-inf"),
    float("inf")
)

print("--------------------------")
print("Optimal value at root:", result)
```
