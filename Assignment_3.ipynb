```python

# Define a function to implement alpha-beta pruning
def alpha_beta(node, level, alpha, beta):
    # Indent for organized printing
    indent = "  " * (level - 1)

    # Number indicates a leaf node, return the value
    if isinstance(node, int):
        print(f"{indent}Leaf: {node}")
        return node

    # Odd levels are MAX nodes.
    if level % 2 == 1:

        print(f"{indent}MAX node (α={alpha}, β={beta})")

        # Initialize starting value to -infinity
        best_value = float("-inf")

        for child in node:

            # Recursively evaluate the child nodes
            successor_val = alpha_beta(
                child,
                level + 1,
                alpha,
                beta
            )

            # Update to best value found so far
            best_value = max(best_value, successor_val)

            # Update alpha based on current values of alpha and best_value
            alpha = max(alpha, best_value)

            print(
                f"{indent}Updated MAX node value={best_value}, α={alpha}"
            )

            # Pruning condition:
            if alpha >= beta:
                print(
                    f"{indent}PRUNE! α = {alpha}, β = {beta}"
                )
                break #Break if pruned, no need to evaluate additional child nodes

    # Even levels are MIN nodes.
    else:

        print(f"{indent}MIN node (α={alpha}, β={beta})")

        # Initialize starting value to +infinity
        best_value = float("inf")

        for child in node:

            # Recursively evaluate the child nodes
            successor_val = alpha_beta(
                child,
                level + 1,
                alpha,
                beta
            )

            # Update best value found so far.
            best_value = min(best_value, successor_val)

            # Update beta based on current values of beta and best_value
            beta = min(beta, best_value)

            print(
                f"{indent}Updated MIN node value={best_value}, β={beta}"
            )

            # Pruning condition:
            if beta <= alpha:
                print(
                    f"{indent}PRUNE! α = {alpha}, β = {beta}"
                )
                break #Break if pruned, no need to evaluate additional child nodes

    # Return the best value found for this subtree.
    return best_value

# Define leaf values of game tree
tree_values = [
    [[3, 5], [6, 9]],
    [[1, 2], [7, 8]]
]

# Apply alpha-beta pruning
result = alpha_beta(
    tree_values,
    level=1,
    alpha=float("-inf"),
    beta=float("inf")
)

# Print value of optimal decision
print(f"\nOptimal: {result}")
```
